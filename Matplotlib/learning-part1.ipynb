```python
# importing matplotlib
import matplotlib.pyplot as plt

# Basic structure using matplotlib
x = [1,2,3,4,5]
y = [2,3,6,8,10]

plt.plot(x,y)
plt.xlabel("X-axis (x marks)")
plt.ylabel("Y-axid (y marks)")
plt.title("X and Y values")
plt.grid(True)                  # This will show the grid in graph
plt.show()
```

```python
# Styling a Line Plot

x = [1,2,3,4,5]
y = [5,6,9,8,7]

plt.plot(x,y,
         color = "blue",
         linestyle = "--",
         linewidth = "2",
         marker = "o",
         markersize = 6
         )

plt.xlabel("Years")
plt.ylabel("Values")
plt.title("Values and Years")
plt.grid(True)
plt.show()
```

```python
# Multiple lines on the same plot

x = [1,2,3,4,5]
y = [5,6,9,8,7]
y2 = [1,5,10,6,8]

plt.plot(x,y, label = "line1")
plt.plot(x,y2, label = "line2")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Multi lines on same plot")
plt.legend()                         # Give the name to the lines
plt.show()
```

```python
# Bat chat
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,9,6,8,10]

plt.bar(x,y)
plt.xlabel("x axis label")
plt.ylabel("y axis label")
plt.title("Bar graph")
plt.grid()
plt.show
```

```python
# Styling a Bar Chart

x = [1,2,3,4,5]
y = [2,9,6,8,10]

plt.bar(x,y,
         color = "skyblue",
         linewidth = 1.5,
         edgecolor = "black",
         width = 0.6 
         )

plt.text(3,6, "Hello")       # It take a argument a first x and sencond y and with double inverted comma where we placa a string or data
plt.show()
```

```python
# Horizontal Bar Chat

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.barh(x,y)
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Horizontal Bar Graph")
plt.grid(True)
plt.show()
```

```python
# Multiple Grouped Bar Charts (Grouped Bars)

import numpy as np
import matplotlib.pyplot as plt

x = np.arange(4)
y1 = [10,15,7,12]
y2 = [8,14,9,10]

width = 0.35

plt.bar(x - width/2,y1,width,label = "Group1")
plt.bar(x + width/2,y2,width,label = "Group2")

plt.xticks(x,["A","B","C","D"])
plt.xlabel("Catogeries")
plt.ylabel("Values")
plt.legend()
plt.show()
```
