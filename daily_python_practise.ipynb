```python
# 1 Print your name, age, city, and a short introduction using print()

dict = {
   "01":{
       "name":"aditya",
       "age":23,
       "city":"Thane"
   } 
}

print(dict["01"])
print(f"Name:", dict["01"]["name"])
print(f"Age:", dict["01"]["age"])
print(f"City:", dict["01"]["city"])
```

```python
# 2. Create variables for an integer, float, string, and boolean and print their values and types.

dict = {
   "01":{
       "name":"aditya",
       "age":23,
       "salary":125000.25,
       "isAlive":True,
       "city":"Thane"
   } 
}

print(type(dict["01"]["age"]))
print(type(dict["01"]["salary"]))
print(type(dict["01"]["isAlive"]))
print(type(dict["01"]["name"]))
```

```python
# 3. Take two numbers as input and print their sum, difference, product, and quotient.

num1 = int(input("Enter Your number 1: "))
num2 = int(input("Enter your number 2: "))

sum = num1 + num2
diff = num2 - num1
product = num1*num2

print("Number1: ",num1)
print("Number2: ",num2)
print(f"The addition of the numbers is: ",sum)
print(f"The product of the Numbers is: ",product)
print(f"The diff of the numbers is: ",diff)
```

```python
# 4. Convert a string number such as '25' into an integer and add 10.

string = "25"
add = 10

number = int(string)
adds = add + number
print("The string is: ",number)
print("The add & total int is: ", adds)
```
