```python
customer = input("Enter Your Name: ")

def menu():
    print("1. Veg Combo = 500")
    print("2. Non-Veg Combo = 550")


def veg():
    gro_veg = 500
    return gro_veg


def non_veg():
    non_vegs = 550
    return non_vegs


def list_input():
    grocery = int(input("Choose one: "))

    if grocery == 1:
        quantity = int(input("Enter quantity: "))
        amount = quantity * veg()
        item = "Veg Combo"

    elif grocery == 2:
        quantity = int(input("Enter quantity: "))
        amount = quantity * non_veg()
        item = "Non-Veg Combo"

    else:
        print("Invalid Choice!")
        return None

    return item, quantity, amount


def output():
    menu()

    result = list_input()

    if result:
        item, quantity, amount = result

        print("====== Summary ======")
        print(f"Name: {customer}")
        print(f"Item: {item}")
        print(f"Quantity: {quantity}")
        print(f"Amount: ₹{amount}")


output()
```
