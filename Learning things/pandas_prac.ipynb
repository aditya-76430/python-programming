```python
# boolean fileatring
import pandas as pd

datai = {
    "ID":[101,102,103,104,105,106,107,108,109,110],
    "name":["aditya","rohan","sohan","mohan","Pohan","tohan","abhishek","sudish","nitish","ritnesh"],
    "department":["Manager","Finance","HR","Accoutant","Finance","HR","Accoutant","Finance","HR","Finance"],
    "salary":[125000,120000,150000,110000,100000,105000,90000,85000,80000,75000]
}

df = pd.DataFrame(datai)
print(df)
```

```python
# From above table

print(df[(df["salary"]>104000) & (df["department"]=="Finance")])
print("========== Finance Dep ===========")
print(df[df["department"]=="Finance"])
print("=========== Hr Dept ==============")
print(df[df["department"]=="HR"])
```

```python
# query like a sql method

df.query("salary > 104000 and department == 'HR'")
```

```python
# Convertinf string into int
import pandas as pd

datas = {
    "ID":[101,102,103,104,105,106,107,108,109,110],
    "name":["aditya "," rohan","soh an","mo han","Pohan","Tohan","Abhishek","sudish","nitish","ritnesh"],
    "department":["Manager","Finance","HR","Accoutant","Finance","HR","Accoutant","Finance","HR","Finance"],
    "salary":[125000,120000,150000,110000,100000,105000,90000,85000,80000,75000]
}

df = pd.DataFrame(datas)
print(df)
```

```python
# changing data types

df[df["salary"] == df["salary"].astype(int)]
print(type(df["salary"]))
```

```python
# Convert to lowercase

df["name"]=df["name"].str.lower()
df["name"]=df["name"].str.capitalize()
df["name"]=df["name"].str.strip()
df["name"]=df["name"].str.split(" ")
#df["name"]=df["name"].str.contains("name")
print(df)
```
