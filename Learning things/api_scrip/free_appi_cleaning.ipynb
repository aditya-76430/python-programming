```python
import requests
import pandas as pd
```

```python
URL = "https://dogapi.dog/api/v2/breeds"

res = URL
res = requests.get(URL)
json_data = res.json()

#Now pandas
df = pd.json_normalize(json_data["data"])
df = df[['attributes.name','type','attributes.male_weight.max','attributes.traits.apartment_friendly']]
df = df.sort_values("attributes.traits.apartment_friendly", ascending = False)
df = df.reset_index(drop = True)
df = df.fillna(0)
df
```

```python
# save this file

df.to_csv("cleaned_dog_data.csv", index=False)
print("CSV file created successfully!")
```
