```python
# 10. Create a menu-driven calculator using conditions

num1 = int(input("Enter Your number: "))
num2 = int(input("Enter Your number: "))

def menu():
    print("1.Add")
    print("2.Sub")
    print("3.Multi")
    
def add():
    return num1 + num2

def sub():
    return num2 - num1

def multi():
    return num1 * num2

while True:
    menu()
    
    choice = int(input("Enter your choice: "))
    
    if choice == 1:
        addition = add()
        print(f"The addition of number {num1} and {num2}is: {addition}")
        
    
    elif choice == 2:
        subtraction = sub()
        print(f"The subtaction of the given numbers {num2} and {num1} is: {subtraction}")    
        
    
    elif choice == 3:
        multiplication = multi()
        print(f"The multiplication of the numbers {num1} and {num2}is: {multiplication}")
        
        
    else:
        print("Please choice the correct number only !")
    
    break
```

```python
# 5. Classify a triangle as equilateral, isosceles, or scalene.

side1 = int(input("Enter side1: "))
side2 = int(input("Enter side2: "))
side3 = int(input("Enter side3: "))

a = side1
b = side2
c = side3

# First check whether the sides can form a triangle
if a + b > c and b + c > a and c + a > b:

    # Equilateral: all 3 sides are equal
    if a == b == c:
        print("Given triangle is an equilateral triangle")

    # Isosceles: any 2 sides are equal
    elif a == b or b == c or a == c:
        print("Given triangle is an isosceles triangle")

    # Scalene: all 3 sides are different
    else:
        print("Given triangle is a scalene triangle")

else:
    print("These sides cannot form a triangle")
    
    
#  Can these sides form a triangle?
#            ↓
#          YES
#            ↓
#     Are all 3 equal?
#       ↙          ↘
#     YES           NO
#      ↓             ↓
# Equilateral    Are any 2 equal?
#                 ↙       ↘
#               YES        NO
#                ↓          ↓
#           Isosceles     Scalene  
```
