```python
# 5. Classify a triangle as equilateral, isosceles, or scalene.

side1 = int(input("Enter side1: "))
side2 = int(input("Enter side2: "))
side3 = int(input("Enter side3: "))

a = side1
b = side2
c = side3

# First check whether the sides can form a triangle
if a + b > c and b + c > a and c + a > b:

    # Equilateral: all 3 sides are equal
    if a == b == c:
        print("Given triangle is an equilateral triangle")

    # Isosceles: any 2 sides are equal
    elif a == b or b == c or a == c:
        print("Given triangle is an isosceles triangle")

    # Scalene: all 3 sides are different
    else:
        print("Given triangle is a scalene triangle")

else:
    print("These sides cannot form a triangle")
    
    
#  Can these sides form a triangle?
#            ↓
#          YES
#            ↓
#     Are all 3 equal?
#       ↙          ↘
#     YES           NO
#      ↓             ↓
# Equilateral    Are any 2 equal?
#                 ↙       ↘
#               YES        NO
#                ↓          ↓
#           Isosceles     Scalene  
```

```python
# 4. Keep taking numbers until the user enters 0, then print the total.

total = 0

while True:
    num = int(input("Enter a number: "))

    if num == 0:
        break

    total += num

print("Total:", total)
```

```python
# 8. Create an ATM menu that continues until the user chooses exit.

balance = 100

def add_money():
    global balance
    
    Enter_amount = int(input("Enter the amount: "))
    
    if Enter_amount > 0:
        global balance
        
        balance = Enter_amount + balance
        print("Amount added sucessfull!", balance)
    else:
         print("Enter Valid Amount !")
       
    return balance


def withdraw_money():
    global balance
    
    Enter_withdraw_amount = int(input("Enter amount: "))
     
    if Enter_withdraw_amount <= balance:
         balance = balance - Enter_withdraw_amount
         print("Amount withdraw sucessfull !", balance)
    else:
        print("Enter valid amount !")
        
    return balance
        
while True:
    
    print("====== ATM ========")
    print("1. Add amount")
    print("2. Withdraw Amount")
    print("3. Check Balance")
    print("4. Exit")
    
    choice = int(input("Enter the option: "))
    
    if choice == 1:
        print("Enter Amount to add: ")
        add_money()
        
    elif choice == 2:
        print("Enter withdrawn amount: ")
        withdraw_money()
        
    elif choice == 3:
        print("Available Balance is: ", balance)
            
    elif choice == 4:
        break
    
    else:
        print("Enter valid choice !")
```
